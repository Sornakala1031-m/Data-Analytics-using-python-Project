```python
import pandas as pd

employees = {
    "Emp_ID": [101, 102, 103, 104, 105],
    "Name": ["Arun", "Priya", "Rahul", "Divya", "Kumar"],
    "Department": ["IT", "HR", "IT", "Finance", "HR"],
    "Salary": [55000, 40000, 50000, 65000, 42000],
    "Experience": [2, 3, 5, 6, 4]
}

df = pd.DataFrame(employees)

print("Employee Dataset:")
print(df)

print("\nAverage Salary:")
print(df["Salary"].mean())

print("\nHighest Salary:")
print(df["Salary"].max())

print("\nEmployees in HR:")
print(df[df["Department"] == "HR"])

print("\nEmployees with Experience > 3:")
print(df[df["Experience"] > 3])

print("\nEmployee Count by Department:")
print(df["Department"].value_counts())
```

```python
import pandas as pd
import numpy as np

sales_data = {
  "Order_ID": [101, 102, 103, 104, 104, 105, 106, 107],
  "Customer": ["John", "Alice", np.nan, "David", "David", "Emma", "Chris", np.nan],
  "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Laptop", np.nan, "Phone", "Tablet"],
  "Quantity": [2, 1, np.nan, 3, 3, 2, 1, 4],
  "Price": [50000, 20000, 15000, np.nan, np.nan, 30000, 20000, 15000]
}
df = pd.DataFrame(sales_data)

print("===== ORIGINAL DATASET =====")
print(df)

print("\n===== DUPLICATE ROWS =====")
print(df.duplicated())

print("\nTotal Duplicates:", df.duplicated().sum())

df = df.drop_duplicates()

print("\n===== AFTER REMOVING DUPLICATES =====")
print(df)

print("\n===== MISSING VALUES =====")
print(df.isnull().sum())

df["Customer"] = df["Customer"].fillna("Unknown")
df["Product"] = df["Product"].fillna("Not Available")
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())
df["Price"] = df["Price"].fillna(df["Price"].mean())

print("\n===== AFTER FILLNA() =====")
print(df)

df["Total_Sales"] = df["Quantity"] * df["Price"]

print("\n===== FINAL CLEANED DATASET =====")
print(df)

df2 = pd.DataFrame(sales_data)

print("\n===== DATASET BEFORE DROPNA() =====")
print(df2)

df2 = df2.dropna()

print("\n===== DATASET AFTER DROPNA() =====")
print(df2)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales_data.csv")

print("\n===== SALES DATA =====")
print(df)

print("\n===== REGION-WISE TOTAL SALES =====")
region_sales = df.groupby("Region")["Sales"].sum()
print(region_sales)

print("\n===== REGION-WISE AGGREGATION REPORT =====")

region_report = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Maximum_Sales=("Sales", "max"),
    Minimum_Sales=("Sales", "min"),
    Total_Quantity=("Quantity", "sum"),
    Average_Quantity=("Quantity", "mean")
)

print(region_report)

print("\n===== REGION & PRODUCT SALES =====")

region_product = df.groupby(
    ["Region", "Product"]
)["Sales"].sum()

print(region_product)

best_region = df.groupby("Region")["Sales"].sum().idxmax()

print("\nBest Performing Region :", best_region)

print("\n===== SALES CONTRIBUTION (%) =====")

percentage = (region_sales / region_sales.sum()) * 100

print(percentage.round(2))

print("\n===== PIVOT TABLE =====")

pivot_table = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Product",
    aggfunc="sum",
    fill_value=0
)

print(pivot_table)


plt.figure(figsize=(8,5))
region_sales.plot(
    kind="bar",
    color=["red","green","blue","orange"]
)

plt.title("Region-wise Total Sales")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.xticks(rotation=0)
plt.show()

plt.figure(figsize=(6,6))

region_sales.plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.title("Sales Percentage by Region")
plt.ylabel("")
plt.show()

plt.figure(figsize=(8,5))

region_sales.sort_values().plot(
    kind="barh",
    color="skyblue"
)

plt.title("Region Sales Ranking")
plt.xlabel("Sales")
plt.show()

pivot_table.plot(
    kind="bar",
    stacked=True,
    figsize=(10,6)
)

plt.title("Region vs Product Sales")
plt.xlabel("Region")
plt.ylabel("Sales")
plt.legend(title="Product")
plt.show()

print("\n===== REPORT GENERATED SUCCESSFULLY =====")
```
